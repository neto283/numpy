```python
lista = [[1, 2, 3, 4],
         [3, 4, 5, 6],
         [5, 6, 7, 8]]

array_2d = np.array(lista, dtype='float')
array_2d
```

## Operaciones vectorizadas

## Generación de matrices aleatorias con Python

```python
def generar_matriz(N):
    matriz = []
    for i in range(N):
        fila = []
        for j in range(N):
            fila.append(random.randint(1, 9))
        matriz.append(fila)
    return matriz

def imprimir_matriz(matriz):
    for fila in matriz:
        print(' '.join(f'{elemento:2}' for elemento in fila))

m_5 = generar_matriz(5)
imprimir_matriz(m_5)
```

```python
m_10 = generar_matriz(10)
m_20 = generar_matriz(20)

print("Matriz 5x5:", len(m_5), "filas x", len(m_5[0]), "columnas")
print("Matriz 10x10:", len(m_10), "filas x", len(m_10[0]), "columnas")
print("Matriz 20x20:", len(m_20), "filas x", len(m_20[0]), "columnas")
```

## La misma idea usando NumPy

```python
matriz_numpy = np.random.randint(1, 10, size=(5, 5))
matriz_numpy
```

### Conclusión

NumPy permite representar datos con arreglos multidimensionales y consultar fácilmente su dimensión, forma, tamaño y tipo de dato. Además, las operaciones vectorizadas simplifican los cálculos sobre matrices.
